```python
import mlflow
import pandas as pd
import matplotlib.pyplot as plt

client = mlflow.tracking.MlflowClient()

exp = client.get_experiment_by_name("Hand_Gesture_Comparison")  

if exp is not None:
    runs = client.search_runs(exp.experiment_id, order_by=["metrics.test_f1 DESC"])
    
    # FIXED: Use run_name instead of model_type parameter
    df_runs = pd.DataFrame([{
        "run_id": r.info.run_id,
        "model_name": r.info.run_name,  # Changed from model_type
        "test_f1": float(r.data.metrics["test_f1"]),
        "test_acc": float(r.data.metrics["test_accuracy"])
    } for r in runs])
    
    # Create chart
    plt.figure(figsize=(8,5))
    df_runs.set_index("model_name")["test_f1"].plot.bar()
    plt.ylabel("Test F1")
    plt.title("Model Comparison (Test F1)")
    plt.tight_layout()
    plt.savefig("reports/model_comparison.png")
    plt.show()
else:
    print("Experiment not found! Run the training script first.")
```
